```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Replace with your file path
file_path = "marketing_campaign.csv"

# Load the CSV file into a DataFrame
df = pd.read_csv(file_path)

# Confirm the dataset is loaded
print("Dataset loaded successfully.")
```

```python
# Display the first 5 rows
print("First few rows of the dataset:")
print(df.head())
```

```python
# Get the shape of the DataFrame
rows, columns = df.shape
print(f"\nNumber of rows: {rows}")
print(f"Number of columns: {columns}")
```

```python
# Get a summary of the dataset
print("\nSummary of the dataset:")
df.info()
```

```python
# Get a summary of the dataset
print("\nSummary of the dataset:")
df.info()
```

```python
# Generate descriptive statistics
print("\nDescriptive statistics for numerical columns:")
print(df.describe())
```

```python
# Number of unique Campaign_ID values
unique_campaigns = df['Campaign_ID'].nunique()
print(f"Number of unique Campaign_ID values: {unique_campaigns}")
```

```python
# Unique values in Location column
unique_locations = df['Location'].unique()
print("\nUnique values in Location column:")
print(unique_locations)

# Unique values in Customer_Segment column
unique_customer_segments = df['Customer_Segment'].unique()
print("\nUnique values in Customer_Segment column:")
print(unique_customer_segments)
```

```python
# Count occurrences in Campaign_Type column
campaign_type_counts = df['Campaign_Type'].value_counts()
print("\nOccurrences of each category in Campaign_Type column:")
print(campaign_type_counts)

# Count occurrences in Channel_Used column
channel_used_counts = df['Channel_Used'].value_counts()
print("\nOccurrences of each category in Channel_Used column:")
print(channel_used_counts)
```

```python
# Scatter plot to visualize relationship between Acquisition_Cost and ROI
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Acquisition_Cost', y='ROI', color='b')
plt.title('Scatter Plot of Acquisition_Cost vs ROI')
plt.xlabel('Acquisition Cost')
plt.ylabel('ROI')
plt.show()
```

```python
# Bar chart to visualize average Conversion_Rate by Channel_Used and Campaign_Type
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Channel_Used', y='Conversion_Rate', hue='Campaign_Type', ci=None)
plt.title('Average Conversion Rate by Channel and Campaign Type')
plt.xlabel('Channel Used')
plt.ylabel('Average Conversion Rate')
plt.show()
```

```python
# Box plot to visualize the distribution of Engagement_Score across Campaign_Type
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Campaign_Type', y='Engagement_Score')
plt.title('Distribution of Engagement Score by Campaign Type')
plt.xlabel('Campaign Type')
plt.ylabel('Engagement Score')
plt.show()
```

```python
# Bar chart to compare the average ROI by Company
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Company', y='ROI', ci=None)
plt.title('Average ROI by Company')
plt.xlabel('Company')
plt.ylabel('Average ROI')
plt.show()
```

```python
# Correlation heatmap between Engagement_Score and Conversion_Rate
correlation_data = df[['Engagement_Score', 'Conversion_Rate']].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_data, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation between Engagement Score and Conversion Rate')
plt.show()
```

```python
# Count plot to visualize the distribution of Target_Audience
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Target_Audience', palette='viridis')
plt.title('Distribution of Target Audience')
plt.xlabel('Target Audience')
plt.ylabel('Count')
plt.xticks(rotation=45)  # Rotate x labels for better visibility
plt.show()
```

```python
# Bar chart to identify highest Conversion_Rate by Customer_Segment and Language
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Language', y='Conversion_Rate', hue='Customer_Segment', ci=None)
plt.title('Average Conversion Rate by Customer Segment and Language')
plt.xlabel('Language')
plt.ylabel('Average Conversion Rate')
plt.show()
```

```python
# Box plot to visualize Acquisition Cost across Customer Segment and Channel Used
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Customer_Segment', y='Acquisition_Cost', hue='Channel_Used')
plt.title('Distribution of Acquisition Cost by Customer Segment and Channel Used')
plt.xlabel('Customer Segment')
plt.ylabel('Acquisition Cost')
plt.show()
```

```python
# Bar chart to analyze average Conversion Rate by Language
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Language', y='Conversion_Rate', ci=None)
plt.title('Average Conversion Rate by Language')
plt.xlabel('Language')
plt.ylabel('Average Conversion Rate')
plt.show()
```

```python
# Bar chart to compare Engagement Score by Channel Used, segmented by Campaign Type
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Channel_Used', y='Engagement_Score', hue='Campaign_Type', ci=None)
plt.title('Engagement Score by Channel Used, Segmented by Campaign Type')
plt.xlabel('Channel Used')
plt.ylabel('Average Engagement Score')
plt.show()
```

```python
# Calculate the total ROI by Channel Used
roi_by_channel = df.groupby('Channel_Used')['ROI'].sum()

# Pie chart for ROI distribution by Channel Used
plt.figure(figsize=(8, 8))
roi_by_channel.plot(kind='pie', autopct='%1.1f%%', startangle=90, cmap='Set3')
plt.title('Distribution of Total ROI by Channel Used')
plt.ylabel('')  # Hide the y-axis label for a cleaner look
plt.show()
```

```python
# Scatter plot to show relationship between Clicks and Impressions, segmented by Channel Used
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Clicks', y='Impressions', hue='Channel_Used', palette='Set2')
plt.title('Relationship between Clicks and Impressions for each Channel Used')
plt.xlabel('Clicks')
plt.ylabel('Impressions')
plt.show()
```

```python
# Histogram to visualize the distribution of Duration
plt.figure(figsize=(10, 6))
sns.histplot(df['Duration'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Duration')
plt.xlabel('Duration')
plt.ylabel('Frequency')
plt.show()
```

```python

# Ensure 'Date' is in datetime format if it's not already
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y', errors='coerce')
# format='%d-%m-%Y' specifies the actual format of your dates
# errors='coerce' handles any invalid dates by setting them to NaT (Not a Time)

# Line chart to analyze Conversion Rate over Date for each Company
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='Date', y='Conversion_Rate', hue='Company', marker='o')
plt.title('Trend of Conversion Rate over Date by Company')
plt.xlabel('Date')
plt.ylabel('Conversion Rate')
plt.legend(title='Company', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Line chart to examine the trend of Engagement Score over Date
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='Date', y='Engagement_Score', marker='o', color='green')
plt.title('Trend of Engagement Score over Date')
plt.xlabel('Date')
plt.ylabel('Engagement Score')
plt.show()
```

```python
# Bar chart to determine the location with the highest Acquisition Cost
plt.figure(figsize=(12, 6))
location_acquisition_cost = df.groupby('Location')['Acquisition_Cost'].sum().sort_values(ascending=False)
location_acquisition_cost.plot(kind='bar', color='salmon')
plt.title('Total Acquisition Cost by Location')
plt.xlabel('Location')
plt.ylabel('Total Acquisition Cost')
plt.xticks(rotation=45)  # Rotate x labels for better visibility
plt.show()
```

```python
# Bar chart to visualize Conversion Rate by Location, categorized by Target Audience
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Location', y='Conversion_Rate', hue='Target_Audience', ci=None)
plt.title('Conversion Rate by Location, Categorized by Target Audience')
plt.xlabel('Location')
plt.ylabel('Average Conversion Rate')
plt.xticks(rotation=45)  # Rotate x labels for better visibility
plt.show()
```

```python
# Calculate the total ROI by Location
roi_by_location = df.groupby('Location')['ROI'].sum()

# Pie chart to show the proportion of ROI by Location
plt.figure(figsize=(8, 8))
roi_by_location.plot(kind='pie', autopct='%1.1f%%', startangle=90, cmap='Set3')
plt.title('Proportion of ROI by Location')
plt.ylabel('')  # Hide the y-axis label for a cleaner look
plt.show()
```
